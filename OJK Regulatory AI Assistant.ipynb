## Chunking

Kode dibawah ini merupakan proses preprocessing dan chunking dokumen PDF pada pipeline RAG. Dokumen PDF dimuat menggunakan PyPDFLoader, kemudian teks dibagi menjadi beberapa chunk menggunakan RecursiveCharacterTextSplitter dengan ukuran 500 karakter dan overlap 50 karakter. Setiap chunk kemudian dibersihkan dan dilengkapi metadata berupa nama file dan nomor halaman sebelum disimpan ke dalam dataset

```python
# ===== Konfigurasi Text Splitter =====
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)


# ===== Menentukan Direktori Dataset PDF =====
pdf_directory = "/Users/promac/Documents/01_Laskar_AI/02_PROJEK/07_Clinical Assistant/all_pdfs"


# ===== Inisialisasi Penyimpanan Data dan ID Dokumen =====
data = []
doc_id = 0


# ===== Membaca dan Memproses Seluruh File PDF =====
for filename in os.listdir(pdf_directory):

    if filename.lower().endswith(".pdf"):

        pdf_path = os.path.join(pdf_directory, filename)

        print(f"📄 Memproses: {filename}")

        # ===== Memuat Dokumen PDF =====
        loader = PyPDFLoader(pdf_path)
        documents = loader.load()

        # ===== Membagi Dokumen Menjadi Chunk =====
        chunks = text_splitter.split_documents(documents)

        # ===== Memproses Setiap Chunk =====
        for chunk_index, chunk in enumerate(chunks):

            # ===== Membersihkan Teks =====
            text = chunk.page_content.strip()

            # ===== Lewati Chunk Kosong =====
            if len(text) < 50:
                continue

            # ===== Metadata =====
            metadata = {
                "source": filename,
                "page": chunk.metadata.get("page", None),
                "chunk_index": chunk_index,
                "document_type": "PNPK"
            }

            # ===== Simpan Data =====
            data.append(
                (str(doc_id), text, metadata)
            )

            doc_id += 1


print(f"\n✅ Jumlah chunk: {len(data)}")
```

## Proses Generasi Vector Embedding

```python
# ===== Import Library untuk Membuat Embedding =====
from sentence_transformers import SentenceTransformer
import numpy as np


# ===== Memuat Model Sentence Transformer pada CPU =====
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device="cpu"
)


# ===== Mengambil Seluruh Teks dari Dataset =====
texts = [item[1] for item in data]


# ===== Menampilkan Jumlah Teks yang Akan Diproses =====
print("Jumlah text:", len(texts))


# ===== Menghasilkan Vector Embedding dari Seluruh Teks =====
embeddings = embedding_model.encode(
    texts,
    batch_size=8,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)


# ===== Menampilkan Dimensi Hasil Embedding =====
print("Embedding shape:", embeddings.shape)
```

```python
# ===== Import FAISS =====

import faiss


# ===== Mengambil Dimensi Embedding =====

dimension = embeddings.shape[1]

print("Dimensi embedding:", dimension)


# ===== Membuat FAISS Index =====

faiss_index = faiss.IndexFlatIP(dimension)


# ===== Memasukkan Seluruh Embedding ke FAISS =====

faiss_index.add(embeddings)


# ===== Mengecek Jumlah Vector =====

print("✅ FAISS Index berhasil dibuat")
print("Jumlah vector:", faiss_index.ntotal)
```

## Retrieval / Pencarian Dokumen Relevan

```python
# ===== Membuat Fungsi Retrieval Menggunakan FAISS =====

def retrieve(query, top_k=5):

    # ===== Menghasilkan Vector Embedding dari Query =====

    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    # ===== Melakukan Pencarian pada FAISS Index =====

    scores, indices = faiss_index.search(
        query_embedding,
        top_k
    )

    # ===== Menyiapkan List Hasil Retrieval =====

    results = []

    # ===== Mengambil Dokumen Berdasarkan ID Hasil FAISS =====

    for score, doc_id in zip(scores[0], indices[0]):

        if doc_id == -1:
            continue

        doc_id = int(doc_id)

        # ===== Mengambil Teks dan Metadata dari Dataset =====

        _, text, metadata = data[doc_id]

        results.append({
            "id": doc_id,
            "score": float(score),
            "text": text,
            "metadata": metadata
        })

    # ===== Mengembalikan Dokumen yang Paling Relevan =====

    return results
```

## Evaluasi RAG

```python
# ============================================
# ===== DATASET EVALUASI RAG OJK =====
# ============================================

evaluation_data = [
    {
        "question": "Apa kewajiban pelaporan kepada Otoritas Jasa Keuangan?",
        "expected_source": "SE%20OJK%20Nomor%2011%20Tahun%202020.pdf"
    },

    {
        "question": "Kapan ketentuan dalam Surat Edaran Otoritas Jasa Keuangan mulai berlaku?",
        "expected_source": "SE%20OJK%20Nomor%2013%20Tahun%202020.pdf"
    },

    {
        "question": "Berapa lama jangka waktu penetapan hasil penilaian kemampuan dan kepatutan?",
        "expected_source": "SE%20OJK%20Nomor%202%20Tahun%202020.pdf"
    }
]

print("Jumlah pertanyaan evaluasi:", len(evaluation_data))
```

```python
# ============================================
# ===== EVALUASI RETRIEVAL FAISS =====
# ============================================

top_k = 5

correct = 0

for i, item in enumerate(evaluation_data, 1):

    question = item["question"]
    expected_source = item["expected_source"]

    results = retrieve(question, top_k)

    retrieved_sources = [
        result["metadata"]["source"]
        for result in results
    ]

    is_relevant = expected_source in retrieved_sources

    if is_relevant:
        correct += 1

    print(f"\nPertanyaan {i}")
    print("Query:", question)
    print("Dokumen ditemukan:", is_relevant)

    for rank, result in enumerate(results, 1):
        print(
            f"{rank}. "
            f"Score={result['score']:.4f} | "
            f"Source={result['metadata']['source']} | "
            f"Page={result['metadata']['page']}"
        )

# ===== Menghitung Recall@K =====

recall_at_k = correct / len(evaluation_data)

print("\n============================================")
print(f"Recall@{top_k}: {recall_at_k:.4f}")
print("============================================")
```

```python
# ============================================
# ===== EVALUASI RECALL@1, RECALL@3, RECALL@5 =====
# ============================================

# ===== Menentukan Nilai K yang Akan Dievaluasi =====

k_values = [1, 3, 5]

# ===== Menyimpan Jumlah Keberhasilan Setiap K =====

results_recall = {}

for k in k_values:

    correct = 0

    print("\n" + "=" * 80)
    print(f"EVALUASI RECALL@{k}")
    print("=" * 80)

    for i, item in enumerate(evaluation_data, 1):

        question = item["question"]
        expected_source = item["expected_source"]

        # ===== Melakukan Retrieval Menggunakan FAISS =====

        results = retrieve(question, top_k=5)

        # ===== Mengambil Source dari Top-K =====

        retrieved_sources = [
            result["metadata"]["source"]
            for result in results[:k]
        ]

        # ===== Mengecek Apakah Dokumen Target Ditemukan =====

        is_found = expected_source in retrieved_sources

        if is_found:
            correct += 1

        print(f"\nPertanyaan {i}")
        print(f"Query   : {question}")
        print(f"Ditemukan di Top-{k}: {is_found}")

    # ===== Menghitung Recall@K =====

    recall = correct / len(evaluation_data)

    results_recall[k] = recall

    print(f"\nRecall@{k}: {recall:.4f} ({recall * 100:.2f}%)")
```

## Evaluasi Gemini

```python
# ============================================
# ===== GENERATE DAN EVALUASI JAWABAN =====
# ============================================

def generate_answer(question):

    # ===== Melakukan Retrieval Menggunakan FAISS =====

    results = retrieve(question, top_k=5)

    # ===== Membuat Context =====

    context_parts = []

    for rank, result in enumerate(results, 1):

        context_parts.append(
            f"""
[Sumber {rank}]
Dokumen: {result['metadata']['source']}
Halaman: {result['metadata']['page']}

{result['text']}
"""
        )

    context = "\n".join(context_parts)

    # ===== Membuat Prompt =====

    prompt = f"""
Jawablah pertanyaan berdasarkan context berikut.

Gunakan hanya informasi yang terdapat dalam context.
Jika informasi tidak tersedia, katakan bahwa informasi tidak ditemukan.

CONTEXT:
{context}

PERTANYAAN:
{question}
"""

    # ===== Generate Jawaban =====

    response = llm.invoke(prompt)

    return response.content, context
```

```python
# ============================================
# ===== MENYIMPAN DATA UNTUK EVALUASI GEMINI =====
# ============================================

evaluation_gemini = []

for i, item in enumerate(evaluation_data, 1):

    question = item["question"]

    # ===== Generate Jawaban Gemini + Context =====

    answer, context = generate_answer(question)

    # ===== Simpan Data Evaluasi =====

    evaluation_gemini.append({
        "question": question,
        "context": context,
        "answer": answer
    })

    print("\n" + "=" * 80)
    print(f"QUERY {i}")
    print("=" * 80)

    print(question)

    print("\nJAWABAN GEMINI")
    print("=" * 80)

    print(answer)
```

```python
# ============================================
# ===== FUNGSI EVALUASI GEMINI =====
# ============================================

import json
import re


def evaluate_gemini(question, context, answer):

    prompt = f"""
Anda adalah evaluator untuk sistem Retrieval-Augmented Generation (RAG).

Evaluasilah jawaban berdasarkan pertanyaan dan context berikut.

========================
PERTANYAAN
========================
{question}

========================
CONTEXT
========================
{context}

========================
JAWABAN GEMINI
========================
{answer}

========================
METRIK EVALUASI
========================

1. CONTEXT RELEVANCE

Apakah context relevan dengan pertanyaan?

1 = tidak relevan
2 = kurang relevan
3 = cukup relevan
4 = relevan
5 = sangat relevan


2. ANSWER FAITHFULNESS

Apakah informasi dalam jawaban didukung oleh context?

1 = banyak klaim tidak didukung
2 = sebagian besar tidak didukung
3 = sebagian didukung
4 = hampir seluruhnya didukung
5 = seluruh klaim penting didukung context


3. ANSWER RELEVANCE

Apakah jawaban menjawab pertanyaan dengan tepat?

1 = tidak menjawab
2 = sangat sedikit menjawab
3 = cukup menjawab
4 = menjawab dengan baik
5 = sangat tepat dan lengkap


========================
FORMAT OUTPUT
========================

Berikan HANYA JSON berikut:

{{
    "context_relevance": {{
        "score": 1,
        "reason": "alasan singkat"
    }},
    "answer_faithfulness": {{
        "score": 1,
        "reason": "alasan singkat"
    }},
    "answer_relevance": {{
        "score": 1,
        "reason": "alasan singkat"
    }}
}}
"""

    # ===== Memanggil LLM sebagai Evaluator =====

    response = llm.invoke(prompt)

    # ===== Mengambil Content dari Response =====

    result = response.content

    # ===== Jika Content Berupa List =====

    if isinstance(result, list):

        text_parts = []

        for item in result:

            if isinstance(item, str):
                text_parts.append(item)

            elif isinstance(item, dict):

                if "text" in item:
                    text_parts.append(item["text"])

                elif "content" in item:
                    text_parts.append(str(item["content"]))

                else:
                    text_parts.append(str(item))

            else:
                text_parts.append(str(item))

        result = "\n".join(text_parts)

    # ===== Pastikan Result Berupa String =====

    result = str(result).strip()

    # ===== Mengambil JSON =====

    match = re.search(
        r'\{.*\}',
        result,
        re.DOTALL
    )

    if not match:

        print("OUTPUT EVALUATOR:")
        print(result)

        raise ValueError(
            "Evaluator tidak menghasilkan JSON yang valid."
        )

    # ===== Parsing JSON =====

    evaluation = json.loads(match.group())

    return evaluation
```

```python
# ============================================
# ===== EVALUASI 3 QUERY GEMINI =====
# ============================================

evaluation_results = []

for i, item in enumerate(evaluation_gemini, 1):

    question = item["question"]
    context = item["context"]
    answer = item["answer"]

    print("\n" + "=" * 80)
    print(f"EVALUASI QUERY {i}")
    print("=" * 80)

    evaluation = evaluate_gemini(
        question=question,
        context=context,
        answer=answer
    )

    evaluation_results.append({
        "query_id": i,
        "question": question,
        "context_relevance": evaluation["context_relevance"]["score"],
        "answer_faithfulness": evaluation["answer_faithfulness"]["score"],
        "answer_relevance": evaluation["answer_relevance"]["score"]
    })

    print("\nContext Relevance:",
          evaluation["context_relevance"]["score"])

    print("Answer Faithfulness:",
          evaluation["answer_faithfulness"]["score"])

    print("Answer Relevance:",
          evaluation["answer_relevance"]["score"])
```

```python
# ============================================
# ===== RATA-RATA EVALUASI GEMINI =====
# ============================================

avg_context_relevance = sum(
    x["context_relevance"]
    for x in evaluation_results
) / len(evaluation_results)

avg_answer_faithfulness = sum(
    x["answer_faithfulness"]
    for x in evaluation_results
) / len(evaluation_results)

avg_answer_relevance = sum(
    x["answer_relevance"]
    for x in evaluation_results
) / len(evaluation_results)

print("\n" + "=" * 80)
print("HASIL AKHIR EVALUASI GEMINI")
print("=" * 80)

print(f"Context Relevance   : {avg_context_relevance:.2f} / 5")
print(f"Answer Faithfulness : {avg_answer_faithfulness:.2f} / 5")
print(f"Answer Relevance    : {avg_answer_relevance:.2f} / 5")
```

```python
# ============================================
# ===== RETRIEVE() STANDALONE =====
# ===== CLINICAL ASSISTANT KEMENKES =====
# ============================================

def retrieve(question, top_k=5):

    # ========================================
    # 1. MENGUBAH QUERY MENJADI EMBEDDING
    # ========================================

    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )


    # ========================================
    # 2. MEMASTIKAN FORMAT FLOAT32
    # ========================================

    query_embedding = np.asarray(
        query_embedding,
        dtype="float32"
    )


    # ========================================
    # 3. FAISS SIMILARITY SEARCH
    # ========================================

    scores, indices = index_loaded.search(
        query_embedding,
        top_k
    )


    # ========================================
    # 4. MENYIAPKAN HASIL RETRIEVAL
    # ========================================

    results = []


    # ========================================
    # 5. MENGAMBIL DATA DARI HASIL FAISS
    # ========================================

    for score, idx in zip(
        scores[0],
        indices[0]
    ):

        # Abaikan index tidak valid
        if idx == -1:
            continue


        # ====================================
        # Mengambil data berdasarkan posisi
        # vector FAISS
        # ====================================

        doc_id, text, metadata = data_loaded[
            int(idx)
        ]


        # ====================================
        # Menyimpan hasil retrieval
        # ====================================

        results.append({

            "id": doc_id,

            "score": float(score),

            "text": text,

            "metadata": metadata

        })


    # ========================================
    # 6. MENGEMBALIKAN HASIL
    # ========================================

    return results
```
